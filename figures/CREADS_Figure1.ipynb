```python
# Upload and extract the Figure 1 inputs.
from pathlib import Path
import zipfile

ROOT = '/content/creads_repro'
Path(ROOT).mkdir(parents=True, exist_ok=True)
required = {'build_fig1_compact_colab.py',
            'data/MATCHED_340REGIONS_WITH_GDL_2023.xlsx',
            'data/COMPLETE_584REGIONS_WITH_CONNECTIVITY_2023.xlsx'}

def is_figure_bundle(path):
    try:
        with zipfile.ZipFile(path) as archive:
            return required.issubset(archive.namelist())
    except zipfile.BadZipFile:
        return False

candidates = [p for p in Path('/content').glob('*.zip') if is_figure_bundle(p)]
if not candidates:
    from google.colab import files
    print('Upload CREADS_Figure1_inputs.zip.')
    uploaded = files.upload()
    candidates = [Path(p) for p in uploaded if is_figure_bundle(Path(p))]
if not candidates:
    raise FileNotFoundError('The uploaded file is not the Figure 1 input bundle.')
with zipfile.ZipFile(candidates[0]) as archive:
    archive.extractall(ROOT)
print('Figure 1 inputs ready.')
```
